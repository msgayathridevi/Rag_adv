```python
import sqlite3
import json
from datetime import datetime

def create_enterprise_database():
    conn = sqlite3.connect("enterprise.db")
    cursor = conn.cursor()
    
    # Finance table
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS finance (
            id INTEGER PRIMARY KEY,
            department TEXT,
            amount REAL,
            quarter TEXT,
            category TEXT,
            date TEXT
        )
    ''')
    
    finance_data = [
        (1, "Engineering", 500000, "Q3", "Salary", "2024-09-30"),
        (2, "Sales", 300000, "Q3", "Commission", "2024-09-30"),
        (3, "Operations", 150000, "Q3", "Equipment", "2024-09-30"),
    ]
    cursor.executemany(
        'INSERT INTO finance VALUES (?, ?, ?, ?, ?, ?)', 
        finance_data
    )
    
    # HR table
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS employees (
            id INTEGER PRIMARY KEY,
            name TEXT,
            department TEXT,
            salary REAL,
            role TEXT,
            clearance_level TEXT
        )
    ''')
    
    employee_data = [
        (1, "John Doe", "Engineering", 120000, "Senior Engineer", "Level 3"),
        (2, "Jane Smith", "HR", 90000, "HR Manager", "Level 2"),
        (3, "Bob Johnson", "Finance", 110000, "Analyst", "Level 3"),
    ]
    cursor.executemany(
        'INSERT INTO employees VALUES (?, ?, ?, ?, ?, ?)', 
        employee_data
    )
    
    # Operations table
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS inventory (
            id INTEGER PRIMARY KEY,
            item TEXT,
            quantity INTEGER,
            location TEXT,
            status TEXT
        )
    ''')
    
    inventory_data = [
        (1, "Laptop", 50, "Warehouse A", "Active"),
        (2, "Server", 10, "Data Center", "Active"),
        (3, "Monitor", 75, "Warehouse B", "Active"),
    ]
    cursor.executemany(
        'INSERT INTO inventory VALUES (?, ?, ?, ?, ?)', 
        inventory_data
    )
    
    conn.commit()
    conn.close()
    print("✅ Database created successfully")

create_enterprise_database()
```
